# Passo 4 - Análise Final e Insights do Projeto

Este notebook tem como objetivo apresentar uma análise final dos resultados obtidos no desenvolvimento do modelo de previsão de engajamento para o projeto WellMe.

Além da interpretação do desempenho da rede neural MLP, também serão discutidos os principais insights identificados ao longo do projeto, possíveis aplicações práticas dentro do contexto do WellMe, limitações da solução desenvolvida e oportunidades de melhoria futura.

O objetivo desta etapa é conectar os resultados técnicos obtidos com possíveis impactos no contexto de saúde e bem-estar, demonstrando como modelos de Inteligência Artificial podem apoiar estratégias de personalização e incentivo à adoção de hábitos saudáveis.

## Principais Resultados Obtidos

O modelo MLP desenvolvido apresentou bom desempenho na classificação dos níveis de engajamento dos usuários, conseguindo identificar padrões relevantes relacionados ao comportamento de atividade física e tempo sedentário.

A análise da matriz de confusão demonstrou que a maior parte das previsões foi realizada corretamente, especialmente para usuários classificados com alto e baixo engajamento.

Além disso, a curva de perda observada durante o treinamento indicou evolução consistente do aprendizado da rede neural ao longo das iterações.

## Insights Identificados

A análise dos dados permitiu identificar que variáveis relacionadas à atividade física, como número de passos e minutos ativos, apresentam forte relação com o nível de engajamento dos usuários.

Por outro lado, o tempo sedentário demonstrou comportamento inversamente proporcional ao engajamento, indicando que usuários mais sedentários tendem a apresentar menor participação em hábitos saudáveis.

Também foi possível observar que usuários classificados como engajamento médio apresentam características mais próximas das demais categorias, tornando essa classe mais difícil de ser distinguida pelo modelo.

## Possíveis Aplicações no WellMe

Os resultados obtidos pelo modelo permitem identificar usuários com diferentes níveis de engajamento dentro da plataforma, possibilitando futuras estratégias de personalização no contexto do WellMe.

Usuários classificados com baixo engajamento (`Low`) podem receber incentivos específicos para aumentar sua participação nas atividades do aplicativo. Entre as possíveis ações, destacam-se:

- desafios mais simples e progressivos;
- metas personalizadas de acordo com o perfil do usuário;
- notificações motivacionais;
- recompensas por pequenas conquistas;
- missões rápidas e de fácil execução;
- recomendações de hábitos saudáveis adaptadas ao nível de atividade do usuário.

Além disso, o modelo também pode auxiliar na identificação precoce de padrões de desmotivação ou abandono da plataforma, permitindo que o WellMe atue preventivamente para manter o usuário engajado.

Para usuários com alto engajamento (`High`), o aplicativo pode oferecer desafios mais avançados, rankings, metas maiores e novos níveis de gamificação, mantendo o interesse contínuo na utilização da plataforma.

Dessa forma, a utilização da Inteligência Artificial no WellMe não se limita apenas à previsão de comportamento, mas também contribui para a construção de experiências mais personalizadas e alinhadas ao perfil de cada usuário.

## Limitações do Projeto

Apesar dos resultados satisfatórios obtidos no desenvolvimento do modelo, este projeto apresenta algumas limitações importantes.

A principal limitação está relacionada ao uso de um dataset público do Fitbit, já que não há acesso a dados reais do aplicativo WellMe. Dessa forma, os dados utilizados representam uma simulação aproximada do comportamento dos usuários no contexto da aplicação.

Além disso, a variável de engajamento utilizada no treinamento do modelo foi criada artificialmente a partir de regras definidas com base em indicadores de atividade física, tempo sedentário e gasto calórico. Embora essa abordagem permita representar o problema proposto, ela não substitui dados reais de interação dos usuários com a plataforma.

Outra limitação está relacionada à quantidade de variáveis disponíveis no dataset. Informações adicionais, como frequência de acesso ao aplicativo, participação em desafios, tempo de uso e interações com elementos de gamificação, poderiam contribuir para um modelo mais completo e preciso.

Também é importante destacar que o modelo foi desenvolvido com fins acadêmicos e exploratórios, podendo ser aprimorado futuramente com conjuntos de dados maiores, novas variáveis e arquiteturas mais avançadas de redes neurais.

O desenvolvimento deste projeto permitiu aplicar conceitos de Inteligência Artificial e Machine Learning no contexto de saúde e bem-estar, utilizando uma rede neural do tipo MLP para prever o nível de engajamento dos usuários.

A partir da análise dos dados e do treinamento do modelo, foi possível identificar padrões relacionados ao comportamento dos usuários, especialmente em variáveis como atividade física, tempo sedentário e gasto calórico.

Os resultados obtidos demonstraram que o modelo foi capaz de classificar os diferentes níveis de engajamento de forma satisfatória, evidenciando o potencial da utilização de redes neurais em aplicações voltadas à personalização de experiências e incentivo à adoção de hábitos saudáveis.

Além dos aspectos técnicos, o projeto também possibilitou compreender como a Inteligência Artificial pode ser integrada ao contexto do WellMe para apoiar estratégias de gamificação, retenção e personalização da experiência dos usuários.

De forma geral, o projeto atingiu os objetivos propostos e demonstrou como soluções baseadas em dados podem contribuir para o desenvolvimento de aplicações mais inteligentes, interativas e alinhadas às necessidades dos usuários.
